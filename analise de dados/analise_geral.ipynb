# Análise Geral Termo

&ensp; Esta análise tem como objetivo avaliar se a distribuição das letras no jogo [termo](https://term.ooo/) se assemelham a distribuição das palavras na língua portuguesa. Isso está sendo feito com o objetivo de avaliar se as palavras do jogo termo possuem algum padrão ou se são palavras aleatórias da língua portuguesa

## Análise da lingua portuguesa

&ensp; Nesta seção iremos avaliar a quantiadade de vezes que cada letra aparece na língua portugesa utilizando como base o "INSERIR FONTE".

```python
import pandas as pd
import matplotlib.pyplot as plt
import requests
import sys, os
sys.path.append(os.path.abspath(".."))
from backend.app import collections
from backend.services.termo_service import get_termos
```

```python
df_portuguese = pd.read_csv('../dictionary/palavras_filtradas.csv')
```

```python
# Dicionário com todas as letras do alfabeto português
alfabeto = {ch: 0 for ch in list("abcdefghijklmnopqrstuvwxyz")}
```

```python
for index, row in df_portuguese.iterrows():
    for char in row['words']:
        if char in alfabeto:
            alfabeto[char] += 1

df_alfabeto_portugues = pd.DataFrame(list(alfabeto.items()), columns=['letra', 'frequencia'])
df_alfabeto_portugues
```

```python
df_alfabeto_portugues.plot(kind='bar', x='letra', y='frequencia', legend=False)

plt.title('Frequência das letras nas palavras com 5 letras (português)')
plt.xlabel('Letras')
plt.ylabel('Frequência')
plt.show()

top10 = df_alfabeto_portugues.nlargest(10, 'frequencia')

top10.plot(kind='bar', x='letra', y='frequencia', legend=False)

plt.title('Top 10 letras mais frequentes em palavras de 5 letras (português)')
plt.xlabel('Letras')
plt.ylabel('Frequência')
plt.show()
```

```python

alfabeto_termo = {ch: 0 for ch in list("abcdefghijklmnopqrstuvwxyz")}
```

```python
for index, row in df_termo.iterrows():
    for char in row['palavra']:
        if char in alfabeto_termo:
            alfabeto_termo[char] += 1

df_alfabeto_termo = pd.DataFrame(list(alfabeto_termo.items()), columns=['letra', 'frequencia'])
df_alfabeto_termo
```

```python
df_alfabeto_termo.plot(kind='bar', x='letra', y='frequencia', legend=False)

plt.title('Frequência das letras nas palavras com 5 letras (termo)')
plt.xlabel('Letras')
plt.ylabel('Frequência')
plt.show()

top10 = df_alfabeto_termo.nlargest(10, 'frequencia')

top10.plot(kind='bar', x='letra', y='frequencia', legend=False)

plt.title('Top 10 letras mais frequentes em palavras de 5 letras (português)')
plt.xlabel('Letras')
plt.ylabel('Frequência')
plt.show()
```
